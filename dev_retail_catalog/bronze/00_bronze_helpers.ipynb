```python
def read_volume(file_name, file_type, volume_path=volume_path):
    df= spark.read.format(file_type).option('header','true').load(f'{volume_path}/{file_name}.{file_type}')
    df= df.withColumn('_ingestion_timestamp', current_timestamp()).withColumn('_source_file', col("_metadata.file_path"))

    return df
```
