```python
dbutils.widgets.text('Schema', 'silver')
schema_name = dbutils.widgets.get('Schema')
print(f'catalog name: {catalog_name}, schema name: {schema_name}')
```

```python
def read_table(table_name, schema_name = 'bronze'):
    print(f'Reading Table from Bronze : {catalog_name}.{schema_name}.{table_name}')
    return spark.read.table(f'{catalog_name}.{schema_name}.{table_name}')
```

```python
def save_table(df,table_name,catalog_name = catalog_name, schema_name = 'silver'):
    print(f'Saving Table in Silver {catalog_name}.{schema_name}.{table_name}')
    df.write.mode('overwrite').option('overwriteSchema', 'true').option('mergeSchema', 'true').saveAsTable(f'{catalog_name}.{schema_name}.{table_name}')
    print(f'Saved Table {catalog_name}.{schema_name}.{table_name}')
```
